## 1. Criar uma amostra (CSV)
No Colab não temos os arquivos do projeto; criamos um CSV de amostra para praticar a leitura.

```python
texto_csv = """uf,estado,regiao,populacao_2025
AM,Amazonas,Norte,4321616
PA,Pará,Norte,8711196
BA,Bahia,Nordeste,14870907
DF,Distrito Federal,Centro-Oeste,2996899
RJ,Rio de Janeiro,Sudeste,17223547
SP,São Paulo,Sudeste,46081801
RS,Rio Grande do Sul,Sul,11233263
"""

with open("populacao_amostra.csv", "w", encoding="utf-8") as f:
    f.write(texto_csv)
print("CSV criado.")
```

## 2. Ler o CSV como lista de dicionários
Cada **linha** vira um **dicionário**; a **tabela** é uma **lista de dicionários**. Usamos só a biblioteca padrão `csv`.

```python
import csv

def carregar_amostra(caminho):
    linhas = []
    with open(caminho, encoding="utf-8") as arquivo:
        for linha in csv.DictReader(arquivo):        # cada linha já vira um dict
            linha["populacao_2025"] = int(linha["populacao_2025"])  # texto -> inteiro
            linhas.append(linha)
    return linhas

dados = carregar_amostra("populacao_amostra.csv")
dados[:2]
```

## 3. Explorar em Python puro (len, set, sum)

```python
qtd_ufs = len({l["uf"] for l in dados})
qtd_regioes = len({l["regiao"] for l in dados})
pop_total = sum(l["populacao_2025"] for l in dados)
print("UFs:", qtd_ufs, "| Regiões:", qtd_regioes, "| População somada:", f"{pop_total:,}".replace(",", "."))
```
